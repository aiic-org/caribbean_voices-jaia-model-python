```python
import os
import zipfile
import shutil

audio_zip_path = './Audio.zip'
audio_dir = './audio_files'

# create extraction directory
os.makedirs(audio_dir, exist_ok=True)

# extract the zipped audio
with zipfile.ZipFile(audio_zip_path, 'r') as zip_ref:
    zip_ref.extractall(audio_dir)

print(f'[INFO] Audio files extracted to: {audio_dir}')
```

```python
import pandas as pd

train_csv, test_csv = 'Train.csv', 'Test.csv'
train_df, test_df = pd.read_csv(train_csv), pd.read_csv(test_csv)

train_df.head()
```

```python
from datasets import Dataset, DatasetDict, Audio

AUDIO_SR = 16000
MAX_SECONDS = 30

def add_audio_paths(df):
    df = df.copy()
    file_names = []
    transcripts = []
    for idx in range(len(df)):
        id = df['ID'][idx]
        if 'Transcription' not in df:
            transcripts.append("")
        audio_path = os.path.sep.join(['./audio_files', f'{id}.wav'])
        file_names.append(audio_path)
    df['audio'] = file_names
    if 'Transcription' not in df:
        df['Transcription'] = transcripts
    return df

train_df = add_audio_paths(train_df)
test_df = add_audio_paths(test_df)

# convert the csvs to hf datasets
train_ds = Dataset.from_pandas(train_df)
test_ds = Dataset.from_pandas(test_df)

train_ds = train_ds.cast_column('audio', Audio(sampling_rate=AUDIO_SR))
test_ds = test_ds.cast_column('audio', Audio(sampling_rate=AUDIO_SR))
split_ds = train_ds.train_test_split(test_size=0.05, seed=42)
train_ds, val_ds = split_ds["train"], split_ds["test"]

dataset = DatasetDict({
    "train": train_ds,
    "val": val_ds,
    "test": test_ds,
})
```
